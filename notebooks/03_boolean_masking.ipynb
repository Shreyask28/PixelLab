### **Boolean masking in image**

creating image for performing boolean masking on it

```python
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((100, 100, 3), dtype=np.uint8)

img[:, :50] = [100, 150, 200]
img[:, 50:] = [200, 100, 50]

plt.imshow(img)
plt.axis("off")
plt.show()
```

#### performing boolean masking (converting the red channels into 0)

```python
red_channel=img[:,:,0]
mask =red_channel>150

img[mask]=[0,0,0]
plt.imshow(img)
plt.axis("off")
plt.show()

```
