#### **basic code for adjusting brigtness of image**

```python

import numpy as np

pixel1 = np.array([100, 150, 200])

# Increase brightness by 30
brighter_pixel = pixel1 + 30

# Decrease brightness by 30
darker_pixel = pixel1 - 30
print("Original pixel:", pixel1)
print("Brighter pixel:", brighter_pixel)
print("Darker pixel:", darker_pixel)
safe_pixel_bright =np.clip(brighter_pixel, 0, 255)
safe_pixel_dark =np.clip(darker_pixel, 0, 255)

print("after clipping brighter pixel", safe_pixel_bright)
print("after clipping darker pixel", safe_pixel_dark)
```



#### **comparing image datatype**


```python


print("original:", pixel1)
print("data type:", pixel1.dtype)

brighter = pixel1 + 30
print("after adding 30:", brighter)
```

#### **converting** image datatype, adjusting , clipping and re converting it

```python
pixel2 =np.array([15, 80, 120], dtype=np.uint8)
pixel_int =pixel2.astype(np.int16)
pixel_bright =pixel_int+30;
pixel_dark =pixel_int-30;
safe_brighter =np.clip(pixel_bright, 0, 255)
safe_dark =np.clip(pixel_dark, 0, 255)
re_safe_brighter =safe_brighter.astype(np.uint8)
re_safe_dark =safe_dark.astype(np.uint8)

print("original:", pixel2)
print("data type:", pixel2.dtype)

print("converted pixel:", pixel_int)
print("converted brighter pixel:", pixel_bright)
print("converted darker pixel:", pixel_dark)
print("converted safe_brighter pixel:", safe_brighter)
print("converted safe_dark pixel:", safe_dark)
```
